```python
!pip install tensorflow-datasets

import tensorflow_datasets as tfds

(train_data, val_data), info = tfds.load(
    'cats_vs_dogs',
    split=['train[:80%]', 'train[80%:]'],
    as_supervised=True,
    with_info=True
)

print("Dataset ready! ✅")
print("Total images:", info.splits['train'].num_examples)
```

```python
import tensorflow as tf

# Image size aur batch size set karo
IMG_SIZE = 160
BATCH_SIZE = 32

# Images ko resize aur normalize karo
def format_image(image, label):
    image = tf.cast(image, tf.float32)
    image = (image/127.5) - 1
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    return image, label

train_batches = train_data.shuffle(1000).map(format_image).batch(BATCH_SIZE).prefetch(1)
val_batches = val_data.map(format_image).batch(BATCH_SIZE).prefetch(1)

print("Data preprocessing done! ✅")
```

```python
# Step 1 - Google ka pretrained MobileNetV2 load karo
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False,
    weights='imagenet'
)

# Pretrained weights freeze karo
base_model.trainable = False

# Step 2 - Apna classifier upar lagao
model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(1, activation='sigmoid')  # 1 = dog, 0 = cat
])

# Step 3 - Model compile karo
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("Model ready! ✅")
model.summary()
```

```python
EPOCHS = 5

history = model.fit(
    train_batches,
    epochs=EPOCHS,
    validation_data=val_batches
)

print("Training done! ✅")
```

```python
!pip install gradio

import gradio as gr
import numpy as np
import tensorflow as tf

def predict(image):
    # Preprocess the image
    image = tf.cast(image, tf.float32)
    image = (image / 127.5) - 1
    image = tf.image.resize(image, (160, 160))
    image = tf.expand_dims(image, axis=0)  # Add batch dimension

    # Predict
    prediction = model.predict(image)[0][0]

    # Return result
    if prediction > 0.5:
        return f"🐶 Dog ({prediction*100:.1f}% confidence)"
    else:
        return f"🐱 Cat ({(1-prediction)*100:.1f}% confidence)"

# Launch the UI
interface = gr.Interface(
    fn=predict,
    inputs=gr.Image(type="numpy"),
    outputs=gr.Textbox(label="Prediction"),
    title="Cat vs Dog Classifier 🐱🐶",
    description="Upload any image and the model will predict!"
)

interface.launch(share=True)  # share=True gives a public link
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Val Accuracy')
plt.legend()
plt.title('Training Progress')
plt.show()
```
